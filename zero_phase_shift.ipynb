```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
```

the idea of zer-phase-shift filter is that it eliminates the phase shift by applying the filter twice
1- Forward filtering (normal filtering)
2- Time reversal: the filtered signal is flipped in time
3- Backward filtering
4- Time is flipped back

Create simple test signal with zeros

```python
#stacking zeros, cosine wave and zeros signals
data = np.concatenate((np.zeros(100), np.cos(np.linspace(np.pi/2, 5*np.pi/2, 10)), np.zeros(100)) , axis=0)
```

Define a low-pass FIR filter (cutoff at 0.6 * Nyquist freq)

```python
fkern = signal.firwin(51, 0.6)
```

Apply casual filter (normal forward filtering)

```python
fdata = signal.lfilter(fkern, 1, data)
```

Apply zero-phase filtering using forward-backward fitlering

```python
fdata_zero_phase = signal.filtfilt(fkern, 1, data)
```

Plotting

```python
plt.figure(figsize=(10,4))
plt.plot(data, 'k', label='Original Signal')
plt.plot(fdata, 'r', label='Causal Filtered Signal (Phase Shifted)')
plt.plot(fdata_zero_phase, 'b', label='Zero-Phase-Shift Filtered Signal')
plt.legend()
plt.show()
```
